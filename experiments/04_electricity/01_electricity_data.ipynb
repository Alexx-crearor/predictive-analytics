```python
# 1. Import libraries
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 2. Locate electricity dataset
candidates = [
    Path.cwd() / "datasets" / "electricity" / "energy_dataset.csv",
    Path.cwd().parent / "datasets" / "electricity" / "energy_dataset.csv",
    Path.cwd().parent.parent / "datasets" / "electricity" / "energy_dataset.csv",
]

DATA_PATH = next(
    (p.resolve() for p in candidates if p.is_file()),
    None
)

if DATA_PATH is None:
    raise FileNotFoundError("Cannot find energy_dataset.csv")

# 3. Read raw dataset
df = pd.read_csv(DATA_PATH)

# 4. Six electricity generation features used in the paper
FEATURE_COLUMNS = [
    "generation biomass",
    "generation fossil brown coal/lignite",
    "generation fossil gas",
    "generation fossil hard coal",
    "generation fossil oil",
    "generation geothermal",
]

assert all(col in df.columns for col in FEATURE_COLUMNS)

# 5. General information
print("Dataset path:", DATA_PATH)
print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

# 6. Missing values in selected features
print("\nMissing values (selected features):")
print(df[FEATURE_COLUMNS].isna().sum())

print("\nRows containing missing selected features:")
print(df[FEATURE_COLUMNS].isna().any(axis=1).sum())

# 7. Explore the six target features
print("\nFeature statistics:")
display(df[FEATURE_COLUMNS].describe().T)

print("\nFirst 5 rows:")
display(df[["time"] + FEATURE_COLUMNS].head())
```

```python
# 1. Chọn khoảng dữ liệu theo bài báo
START_IDX = 4000
END_IDX = 7500

electricity_df = df.iloc[
    START_IDX:END_IDX
].copy()

# 2. Chỉ giữ 6 nguồn phát điện
electricity_raw = electricity_df[
    FEATURE_COLUMNS
].to_numpy(dtype=np.float32)

# 3. Kiểm tra kích thước
print("Selected DataFrame:", electricity_df.shape)
print("Electricity array:", electricity_raw.shape)

assert electricity_raw.shape == (3500, 6)

# 4. Kiểm tra khoảng thời gian
print("\nStart time:", electricity_df["time"].iloc[0])
print("End time:", electricity_df["time"].iloc[-1])

# 5. Kiểm tra missing values
print("\nMissing values per feature:")
print(electricity_df[FEATURE_COLUMNS].isna().sum())

print("\nRows containing NaN:")
print(np.isnan(electricity_raw).any(axis=1).sum())

# 6. Kiểm tra các cột không thay đổi
print("\nUnique values per feature:")
print(electricity_df[FEATURE_COLUMNS].nunique())

# 7. Hiển thị thống kê của 3.500 dòng
display(
    electricity_df[FEATURE_COLUMNS].describe().T
)
```

```python
# 1. Xác định vị trí dòng bị thiếu
missing_mask = electricity_df[FEATURE_COLUMNS].isna().any(axis=1)

missing_rows = electricity_df.loc[
    missing_mask,
    ["time"] + FEATURE_COLUMNS
]

print("Missing rows:")
display(missing_rows)

# 2. Forward Fill: điền bằng giá trị hợp lệ gần nhất trước đó
electricity_clean = (
    electricity_df[FEATURE_COLUMNS]
    .ffill()
    .copy()
)

# 3. Chuyển sang NumPy array
electricity_raw = electricity_clean.to_numpy(
    dtype=np.float32
)

# 4. Kiểm tra dữ liệu sau xử lý
assert electricity_raw.shape == (3500, 6)
assert np.isfinite(electricity_raw).all()

print("\nAfter preprocessing:")
print("Shape:", electricity_raw.shape)
print("Remaining NaN:", np.isnan(electricity_raw).sum())

# 5. Kiểm tra đặc trưng Geothermal
geothermal_idx = FEATURE_COLUMNS.index(
    "generation geothermal"
)

print("\nGeothermal unique values:")
print(np.unique(electricity_raw[:, geothermal_idx]))

assert np.all(electricity_raw[:, geothermal_idx] == 0)

print("\nElectricity preprocessing completed!")
```

```python
from sklearn.preprocessing import MinMaxScaler
import torch
import hashlib

# 1. Chia Train/Test theo thứ tự thời gian
TRAIN_RATIO = 0.8

n_samples = len(electricity_raw)
split_idx = int(n_samples * TRAIN_RATIO)

train_raw = electricity_raw[:split_idx]
test_raw = electricity_raw[split_idx:]

print("Train raw shape:", train_raw.shape)
print("Test raw shape:", test_raw.shape)

assert train_raw.shape == (2800, 6)
assert test_raw.shape == (700, 6)

# 2. Khởi tạo Min-Max Scaler
scaler = MinMaxScaler(feature_range=(0, 1))

# Chỉ fit trên Train để tránh data leakage
train_scaled = scaler.fit_transform(train_raw)

# Test dùng đúng scaler của Train
test_scaled = scaler.transform(test_raw)

# Chuyển về float32
train_scaled = train_scaled.astype(np.float32)
test_scaled = test_scaled.astype(np.float32)

# 3. Kiểm tra kết quả chuẩn hóa
print("\nAfter Min-Max Scaling:")
print("Train shape:", train_scaled.shape)
print("Test shape:", test_scaled.shape)

print("\nTrain min per channel:")
print(train_scaled.min(axis=0))

print("\nTrain max per channel:")
print(train_scaled.max(axis=0))

print("\nTest min per channel:")
print(test_scaled.min(axis=0))

print("\nTest max per channel:")
print(test_scaled.max(axis=0))

assert np.isfinite(train_scaled).all()
assert np.isfinite(test_scaled).all()

# 4. Tạo hash để kiểm tra dữ liệu khi huấn luyện trên GPU
data_hash = hashlib.sha256(
    electricity_raw.tobytes()
).hexdigest()

# 5. Lưu dữ liệu đã chuẩn bị
project_root = DATA_PATH.parents[2]

EXP_DIR = project_root / "experiments" / "04_electricity"
DATA_DIR = EXP_DIR / "data"

DATA_DIR.mkdir(parents=True, exist_ok=True)

bundle = {
    "train": torch.from_numpy(train_scaled),
    "test": torch.from_numpy(test_scaled),
    "channels": FEATURE_COLUMNS,
    "data_hash": data_hash,
    "train_ratio": TRAIN_RATIO,
    "source_start_idx": START_IDX,
    "source_end_idx": END_IDX,
    "missing_strategy": "forward_fill",
    "scaler_min": scaler.data_min_.tolist(),
    "scaler_max": scaler.data_max_.tolist(),
}

output_path = DATA_DIR / "electricity_train_test.pt"

torch.save(bundle, output_path)

print("\nSaved:", output_path)
print("Data SHA256:", data_hash)
```

```python
# 1. Hàm tạo Sliding Windows
def create_windows(data, patch_len, horizon):

    X, y = [], []

    # Giữ đúng quy tắc tạo window trong code tác giả
    for i in range(len(data) - patch_len - horizon):

        # Input: patch_len timestamps liên tiếp
        x_window = data[i : i + patch_len]

        # Target: horizon timestamps kế tiếp
        y_window = data[
            i + patch_len : i + patch_len + horizon
        ]

        # Flatten theo từng channel
        y_window = y_window.T.reshape(-1)

        X.append(x_window)
        y.append(y_window)

    # Chuyển sang PyTorch Tensor
    X = torch.tensor(
        np.stack(X),
        dtype=torch.float32
    )

    y = torch.tensor(
        np.stack(y),
        dtype=torch.float32
    )

    return X, y


# 2. Chọn một cấu hình để kiểm tra
patch_len = 70
horizon = 2

# 3. Tạo windows riêng cho Train và Test
X_train, y_train = create_windows(
    train_scaled,
    patch_len,
    horizon
)

X_test, y_test = create_windows(
    test_scaled,
    patch_len,
    horizon
)

# 4. Hiển thị kích thước
print("Configuration:")
print("Patch Length:", patch_len)
print("Forecast Horizon:", horizon)

print("\nTrain:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTest:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

# 5. Kiểm tra kích thước
assert X_train.shape == (2728, 70, 6)
assert y_train.shape == (2728, 12)

assert X_test.shape == (628, 70, 6)
assert y_test.shape == (628, 12)

# 6. Kiểm tra cửa sổ đầu tiên
np.testing.assert_allclose(
    X_train[0].numpy(),
    train_scaled[:70]
)

np.testing.assert_allclose(
    y_train[0].numpy(),
    train_scaled[70:72].T.reshape(-1)
)

print("\nAll window checks passed!")

# 7. Minh họa một target vector
print("\nFirst target vector:")
print(y_train[0].numpy())
```

```python
# 1. Grid Search parameters
PATCH_LENGTHS = [70, 80, 90, 100, 110, 120, 130]
HORIZONS = [2, 4, 6, 8, 10, 12, 14, 16]

MODEL_NAMES = [
    "LSTM",
    "PatchTST",
    "Autoformer",
    "Informer"
]

EPOCHS = 4000

# 2. Kiểm tra từng cấu hình
grid_results = []

for p in PATCH_LENGTHS:
    for h in HORIZONS:

        # Tạo Sliding Windows
        X_train, y_train = create_windows(
            train_scaled, p, h
        )

        X_test, y_test = create_windows(
            test_scaled, p, h
        )

        # Số cửa sổ theo công thức của tác giả
        expected_train = len(train_scaled) - p - h
        expected_test = len(test_scaled) - p - h

        # Kiểm tra input và output
        assert X_train.shape == (expected_train, p, 6)
        assert y_train.shape == (expected_train, h * 6)

        assert X_test.shape == (expected_test, p, 6)
        assert y_test.shape == (expected_test, h * 6)

        # Kiểm tra không có NaN hoặc Inf
        assert torch.isfinite(X_train).all()
        assert torch.isfinite(y_train).all()
        assert torch.isfinite(X_test).all()
        assert torch.isfinite(y_test).all()

        # Lưu thống kê
        grid_results.append({
            "PatchLen": p,
            "Horizon": h,
            "TrainWindows": len(X_train),
            "TestWindows": len(X_test),
            "InputShape": str(tuple(X_train.shape[1:])),
            "TargetDim": y_train.shape[1]
        })

# 3. Tổng hợp
grid_df = pd.DataFrame(grid_results)

print("=== ELECTRICITY GRID SEARCH VALIDATION ===")
print("Models:", MODEL_NAMES)
print("Patch Lengths:", PATCH_LENGTHS)
print("Horizons:", HORIZONS)
print("Epochs per run:", EPOCHS)

print("\nUnique (P, H) configurations:", len(grid_df))
print("Total training runs:", len(grid_df) * len(MODEL_NAMES))

assert len(grid_df) == 56
assert len(grid_df) * len(MODEL_NAMES) == 224

# 4. Thống kê các cấu hình
display(grid_df)

print("\nTrain windows range:",
      grid_df["TrainWindows"].min(),
      "to",
      grid_df["TrainWindows"].max())

print("Test windows range:",
      grid_df["TestWindows"].min(),
      "to",
      grid_df["TestWindows"].max())

print("\nAll 56 configurations validated successfully!")
```
