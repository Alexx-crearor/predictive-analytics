# Thí nghiệm 04 — Electricity: Chuẩn bị dữ liệu

## 1. Đọc và kiểm tra dữ liệu nguồn

Notebook chuẩn bị dữ liệu phát điện cho bài toán dự báo chuỗi thời gian đa biến.

Nguồn dữ liệu: `datasets/electricity/energy_dataset.csv`.

Dữ liệu gốc gồm 35.064 quan sát và 29 cột.

Thực nghiệm sử dụng sáu kênh:

1. Biomass
2. Fossil brown coal/lignite
3. Fossil gas
4. Fossil hard coal
5. Fossil oil
6. Geothermal

Quy trình: lựa chọn khoảng dữ liệu, xử lý thiếu, chuẩn hóa, chia Train/Test, tạo sliding windows và kiểm tra toàn bộ cấu hình Grid Search.


```python
from pathlib import Path

import hashlib
import numpy as np
import pandas as pd
import torch

from sklearn.preprocessing import MinMaxScaler


# 1. Tìm repository root
project_root = next(
    (
        p for p in [Path.cwd(), *Path.cwd().parents]
        if (
            p / "datasets" / "electricity"
            / "energy_dataset.csv"
        ).is_file()
    ),
    None
)

if project_root is None:
    raise FileNotFoundError(
        "Không tìm thấy datasets/electricity/energy_dataset.csv"
    )


# 2. Đường dẫn dữ liệu
DATA_PATH = (
    project_root
    / "datasets"
    / "electricity"
    / "energy_dataset.csv"
)

EXP_DIR = (
    project_root
    / "experiments"
    / "04_electricity"
)

DATA_DIR = EXP_DIR / "data"

# Không tạo lại dữ liệu trong quá trình cleanup
REBUILD_DATA_FILES = False


# 3. Các kênh Electricity sử dụng trong thực nghiệm
FEATURE_COLUMNS = [
    "generation biomass",
    "generation fossil brown coal/lignite",
    "generation fossil gas",
    "generation fossil hard coal",
    "generation fossil oil",
    "generation geothermal",
]


# 4. Đọc CSV gốc
df = pd.read_csv(DATA_PATH)

assert df.shape == (35064, 29)
assert all(col in df.columns for col in FEATURE_COLUMNS)


# 5. Kiểm tra dữ liệu
print("=== ELECTRICITY DATASET ===")
print("Source:", DATA_PATH)
print("Shape:", df.shape)

print("\nSelected features:")
for i, feature in enumerate(FEATURE_COLUMNS, start=1):
    print(f"{i}. {feature}")

print("\nMissing values:")
print(df[FEATURE_COLUMNS].isna().sum())

print("\nFirst 5 rows:")
display(df[["time"] + FEATURE_COLUMNS].head())
```

## 2. Lựa chọn khoảng dữ liệu

Dữ liệu thực nghiệm được lấy từ 3.500 dòng liên tiếp trong CSV nguồn:

- Chỉ số bắt đầu: 4000 (bao gồm).
- Chỉ số kết thúc: 7500 (không bao gồm).

Chọn đúng sáu cột phát điện đã xác định.

Ma trận thu được có kích thước:

\[
X\in\mathbb{R}^{3500\times6}
\]

Giữ nguyên thứ tự bản ghi và kiểu dữ liệu `float32`.

```python
# 1. Chọn khoảng dữ liệu theo bài báo
START_IDX = 4000
END_IDX = 7500

electricity_df = df.iloc[
    START_IDX:END_IDX
].copy()

# 2. Chỉ giữ 6 nguồn phát điện
electricity_raw = electricity_df[
    FEATURE_COLUMNS
].to_numpy(dtype=np.float32)

# 3. Kiểm tra kích thước
print("Selected DataFrame:", electricity_df.shape)
print("Electricity array:", electricity_raw.shape)

assert electricity_raw.shape == (3500, 6)

# 4. Kiểm tra khoảng thời gian
print("\nStart time:", electricity_df["time"].iloc[0])
print("End time:", electricity_df["time"].iloc[-1])

# 5. Kiểm tra missing values
print("\nMissing values per feature:")
print(electricity_df[FEATURE_COLUMNS].isna().sum())

print("\nRows containing NaN:")
print(np.isnan(electricity_raw).any(axis=1).sum())

# 6. Kiểm tra các cột không thay đổi
print("\nUnique values per feature:")
print(electricity_df[FEATURE_COLUMNS].nunique())

# 7. Hiển thị thống kê của 3.500 dòng
display(
    electricity_df[FEATURE_COLUMNS].describe().T
)
```

## 3. Xử lý dữ liệu thiếu

Trong 3.500 quan sát được chọn, có một dòng chứa giá trị thiếu ở các đặc trưng sử dụng.

Phương pháp xử lý: forward-fill (`ffill`), thay giá trị thiếu bằng giá trị hợp lệ gần nhất trước đó trong cùng cột.

Sau xử lý:

- Shape: `(3500, 6)`.
- Không còn NaN hoặc Inf.
- Kênh `generation geothermal` có giá trị 0 trên toàn bộ khoảng dữ liệu.

Giữ nguyên kênh geothermal để bảo toàn cấu trúc sáu chiều của thực nghiệm.

```python
# 1. Xác định vị trí dòng bị thiếu
missing_mask = electricity_df[FEATURE_COLUMNS].isna().any(axis=1)

missing_rows = electricity_df.loc[
    missing_mask,
    ["time"] + FEATURE_COLUMNS
]

print("Missing rows:")
display(missing_rows)

# 2. Forward Fill: điền bằng giá trị hợp lệ gần nhất trước đó
electricity_clean = (
    electricity_df[FEATURE_COLUMNS]
    .ffill()
    .copy()
)

# 3. Chuyển sang NumPy array
electricity_raw = electricity_clean.to_numpy(
    dtype=np.float32
)

# 4. Kiểm tra dữ liệu sau xử lý
assert electricity_raw.shape == (3500, 6)
assert np.isfinite(electricity_raw).all()

print("\nAfter preprocessing:")
print("Shape:", electricity_raw.shape)
print("Remaining NaN:", np.isnan(electricity_raw).sum())

# 5. Kiểm tra đặc trưng Geothermal
geothermal_idx = FEATURE_COLUMNS.index(
    "generation geothermal"
)

print("\nGeothermal unique values:")
print(np.unique(electricity_raw[:, geothermal_idx]))

assert np.all(electricity_raw[:, geothermal_idx] == 0)

print("\nElectricity preprocessing completed!")
```

## 4. Train/Test Split, chuẩn hóa và kiểm tra dữ liệu

### 4.1. Chia dữ liệu

Dữ liệu được chia theo thứ tự thời gian, không shuffle.

| Tập | Số quan sát | Số kênh |
|---|---:|---:|
| Train | 2800 | 6 |
| Test | 700 | 6 |

### 4.2. Min-Max Scaling

Sử dụng `MinMaxScaler(feature_range=(0, 1))`.

Scaler chỉ được khớp trên Train và sử dụng lại cho Test để tránh data leakage.

Kênh geothermal là kênh hằng và vẫn được giữ lại.

### 4.3. Fingerprint

SHA-256 được tính từ **ma trận dữ liệu gốc đã forward-fill**, trước khi chia Train/Test và trước khi chuẩn hóa.

Điều này khác quy tắc tính hash của thí nghiệm Financial.

Fingerprint kỳ vọng:

`c024e2bb9460bf6e35d6f3353b82b6d90bee422366be0d6b3bb769a340eb0f0f`

### 4.4. Bảo toàn kết quả

Không ghi lại `electricity_train_test.pt`.

Kiểm tra trực tiếp tensor và metadata của bundle đã lưu để xác nhận pipeline vẫn tạo cùng kết quả.


```python
# 1. Chia tuần tự Train/Test
TRAIN_RATIO = 0.8

split_idx = int(len(electricity_raw) * TRAIN_RATIO)

train_raw = electricity_raw[:split_idx]
test_raw = electricity_raw[split_idx:]

assert train_raw.shape == (2800, 6)
assert test_raw.shape == (700, 6)


# 2. Fit Min-Max Scaler trên Train
scaler = MinMaxScaler(feature_range=(0, 1))

train_scaled = scaler.fit_transform(train_raw)
test_scaled = scaler.transform(test_raw)

train_scaled = train_scaled.astype(np.float32)
test_scaled = test_scaled.astype(np.float32)


# 3. Kiểm tra dữ liệu chuẩn hóa
assert train_scaled.shape == (2800, 6)
assert test_scaled.shape == (700, 6)

assert np.isfinite(train_scaled).all()
assert np.isfinite(test_scaled).all()


# 4. Hash dữ liệu gốc sau forward-fill
data_hash = hashlib.sha256(
    electricity_raw.tobytes()
).hexdigest()

EXPECTED_HASH = (
    "c024e2bb9460bf6e35d6f3353b82b6d90bee"
    "422366be0d6b3bb769a340eb0f0f"
)

assert data_hash == EXPECTED_HASH, (
    "Electricity preprocessing hash mismatch!"
)


# 5. Tạo bundle tương ứng trong bộ nhớ
bundle = {
    "train": torch.from_numpy(train_scaled),
    "test": torch.from_numpy(test_scaled),
    "channels": FEATURE_COLUMNS,
    "data_hash": data_hash,
    "train_ratio": TRAIN_RATIO,
    "source_start_idx": START_IDX,
    "source_end_idx": END_IDX,
    "missing_strategy": "forward_fill",
    "scaler_min": scaler.data_min_.tolist(),
    "scaler_max": scaler.data_max_.tolist(),
}


# 6. Đối chiếu bundle đã lưu
output_path = DATA_DIR / "electricity_train_test.pt"

if not output_path.is_file():
    raise FileNotFoundError(output_path)

# Bundle chỉ chứa tensor và metadata đơn giản
saved = torch.load(
    output_path,
    map_location="cpu",
    weights_only=True
)


# 7. Xác minh metadata
assert saved["data_hash"] == EXPECTED_HASH
assert saved["channels"] == FEATURE_COLUMNS
assert saved["train_ratio"] == TRAIN_RATIO

assert saved["source_start_idx"] == START_IDX
assert saved["source_end_idx"] == END_IDX
assert saved["missing_strategy"] == "forward_fill"


# 8. Xác minh tensor tuyệt đối giống nhau
assert torch.equal(
    saved["train"],
    bundle["train"]
)

assert torch.equal(
    saved["test"],
    bundle["test"]
)


# 9. Xác minh scaler
np.testing.assert_array_equal(
    saved["scaler_min"],
    bundle["scaler_min"]
)

np.testing.assert_array_equal(
    saved["scaler_max"],
    bundle["scaler_max"]
)


# 10. Tổng kết
print("=== ELECTRICITY DATA VALIDATION ===")
print("Train:", tuple(saved["train"].shape))
print("Test:", tuple(saved["test"].shape))
print("Channels:", len(saved["channels"]))

print("\nSHA256:", data_hash)
print("Hash matched:", True)

print("\nTrain min:", train_scaled.min(axis=0))
print("Train max:", train_scaled.max(axis=0))

print("\nTest min:", test_scaled.min(axis=0))
print("Test max:", test_scaled.max(axis=0))

print("\nExisting bundle matches preprocessing.")
print("No dataset or checkpoint overwritten.")
```

## 5. Tạo sliding windows

Với input length \(P\), forecast horizon \(H\) và sáu kênh:

- Input shape: `(N, P, 6)`.
- Target shape: `(N, H × 6)`.

Target được làm phẳng theo từng kênh bằng `y_window.T.reshape(-1)`.

Giữ nguyên quy tắc số cửa sổ:

`range(len(data) - patch_len - horizon)`

Không thêm `+1` để duy trì số lượng cửa sổ của thực nghiệm đã huấn luyện.

Cấu hình kiểm tra: `P=70`, `H=2`.

Kích thước kỳ vọng:

- `X_train`: `(2728,70,6)`.
- `y_train`: `(2728,12)`.
- `X_test`: `(628,70,6)`.
- `y_test`: `(628,12)`.

```python
# 1. Hàm tạo Sliding Windows
def create_windows(data, patch_len, horizon):

    X, y = [], []

    # Giữ đúng quy tắc tạo window trong code tác giả
    for i in range(len(data) - patch_len - horizon):

        # Input: patch_len timestamps liên tiếp
        x_window = data[i : i + patch_len]

        # Target: horizon timestamps kế tiếp
        y_window = data[
            i + patch_len : i + patch_len + horizon
        ]

        # Flatten theo từng channel
        y_window = y_window.T.reshape(-1)

        X.append(x_window)
        y.append(y_window)

    # Chuyển sang PyTorch Tensor
    X = torch.tensor(
        np.stack(X),
        dtype=torch.float32
    )

    y = torch.tensor(
        np.stack(y),
        dtype=torch.float32
    )

    return X, y


# 2. Chọn một cấu hình để kiểm tra
patch_len = 70
horizon = 2

# 3. Tạo windows riêng cho Train và Test
X_train, y_train = create_windows(
    train_scaled,
    patch_len,
    horizon
)

X_test, y_test = create_windows(
    test_scaled,
    patch_len,
    horizon
)

# 4. Hiển thị kích thước
print("Configuration:")
print("Patch Length:", patch_len)
print("Forecast Horizon:", horizon)

print("\nTrain:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTest:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

# 5. Kiểm tra kích thước
assert X_train.shape == (2728, 70, 6)
assert y_train.shape == (2728, 12)

assert X_test.shape == (628, 70, 6)
assert y_test.shape == (628, 12)

# 6. Kiểm tra cửa sổ đầu tiên
np.testing.assert_allclose(
    X_train[0].numpy(),
    train_scaled[:70]
)

np.testing.assert_allclose(
    y_train[0].numpy(),
    train_scaled[70:72].T.reshape(-1)
)

print("\nAll window checks passed!")

# 7. Minh họa một target vector
print("\nFirst target vector:")
print(y_train[0].numpy())
```

## 6. Kiểm tra không gian Grid Search

Các cấu hình:

- Patch lengths: `[70,80,90,100,110,120,130]`.
- Horizons: `[2,4,6,8,10,12,14,16]`.
- Models: LSTM, PatchTST, Autoformer, Informer.
- Epochs: 4000.

Số cặp `(P,H)`:

\[
7\times8=56
\]

Tổng số cấu hình huấn luyện:

\[
56\times4=224
\]

Cell kiểm tra kích thước của sliding windows và sự tồn tại của giá trị NaN/Inf cho từng cặp cấu hình.

Không khởi tạo mô hình hoặc thực hiện Training trong bước này.

```python
# 1. Grid Search parameters
PATCH_LENGTHS = [70, 80, 90, 100, 110, 120, 130]
HORIZONS = [2, 4, 6, 8, 10, 12, 14, 16]

MODEL_NAMES = [
    "LSTM",
    "PatchTST",
    "Autoformer",
    "Informer"
]

EPOCHS = 4000

# 2. Kiểm tra từng cấu hình
grid_results = []

for p in PATCH_LENGTHS:
    for h in HORIZONS:

        # Tạo Sliding Windows
        X_train, y_train = create_windows(
            train_scaled, p, h
        )

        X_test, y_test = create_windows(
            test_scaled, p, h
        )

        # Số cửa sổ theo công thức của tác giả
        expected_train = len(train_scaled) - p - h
        expected_test = len(test_scaled) - p - h

        # Kiểm tra input và output
        assert X_train.shape == (expected_train, p, 6)
        assert y_train.shape == (expected_train, h * 6)

        assert X_test.shape == (expected_test, p, 6)
        assert y_test.shape == (expected_test, h * 6)

        # Kiểm tra không có NaN hoặc Inf
        assert torch.isfinite(X_train).all()
        assert torch.isfinite(y_train).all()
        assert torch.isfinite(X_test).all()
        assert torch.isfinite(y_test).all()

        # Lưu thống kê
        grid_results.append({
            "PatchLen": p,
            "Horizon": h,
            "TrainWindows": len(X_train),
            "TestWindows": len(X_test),
            "InputShape": str(tuple(X_train.shape[1:])),
            "TargetDim": y_train.shape[1]
        })

# 3. Tổng hợp
grid_df = pd.DataFrame(grid_results)

print("=== ELECTRICITY GRID SEARCH VALIDATION ===")
print("Models:", MODEL_NAMES)
print("Patch Lengths:", PATCH_LENGTHS)
print("Horizons:", HORIZONS)
print("Epochs per run:", EPOCHS)

print("\nUnique (P, H) configurations:", len(grid_df))
print("Total training runs:", len(grid_df) * len(MODEL_NAMES))

assert len(grid_df) == 56
assert len(grid_df) * len(MODEL_NAMES) == 224

# 4. Thống kê các cấu hình
display(grid_df)

print("\nTrain windows range:",
      grid_df["TrainWindows"].min(),
      "to",
      grid_df["TrainWindows"].max())

print("Test windows range:",
      grid_df["TestWindows"].min(),
      "to",
      grid_df["TestWindows"].max())

print("\nAll 56 configurations validated successfully!")
```
